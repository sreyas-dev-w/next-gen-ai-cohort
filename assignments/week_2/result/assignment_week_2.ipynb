```python
# Part 1 — Data Understanding & Exploration
```

```python
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
import numpy as np
```

```python
# Read the data from the source using pandas lib

data = pd.read_csv("/content/diabetes.csv")
```

```python
data.head()
```

```python
# Print the shape of the data

print(data.shape)
```

```python
# Print all columns

print(data.columns)
```

```python
# Print the data type of each column

print(data.dtypes)
```

```python
# Print count of null values in each column in the data

print(data.isnull().sum())
```

```python
# Identify duplicate records (if any)

duplicated_records = data.duplicated().sum()
print(f"Total duplicate records: {duplicated_records}")
```

```python
# Displays the rows that are duplicated (if any)

duplicated_rows = data[data.duplicated()]
print(duplicated_rows)
```

```python
# Part 2 - Data Cleaning & Preprocessing
```

```python
# Check for zero values in numerical columns that should not be zero

print("Zero values in each column:")
print((data == 0).sum())
```

```python
# Replace zero values with NaN for columns where zero is unlikely
# Glucose, BloodPressure, SkinThickness, Insulin, BMI should not be zero

cols_with_zeros = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']

for col in cols_with_zeros:
    data[col] = data[col].replace(0, np.nan)
```

```python
# Check missing values after replacement

print('Missing values after zero replacement:')
print(data.isnull().sum())
```

```python
# Fill missing values using mean imputation

data['Glucose'] = data['Glucose'].fillna(data['Glucose'].mean())
data['BloodPressure'] = data['BloodPressure'].fillna(data['BloodPressure'].mean())
data['SkinThickness'] = data['SkinThickness'].fillna(data['SkinThickness'].mean())
data['Insulin'] = data['Insulin'].fillna(data['Insulin'].mean())
data['BMI'] = data['BMI'].fillna(data['BMI'].mean())

# Verify no more missing values
print('Missing values after cleanup:')
print(data.isnull().sum())
```

```python
# Part 3 - Data Analysis
```

```python
# Statistical analysis

data.describe()
```

```python
# Check the distribution of the target variable (Outcome)

outcome_count = data['Outcome'].value_counts()

print('Outcome Distribution:')
print(outcome_count)
print()
print('Outcome Percentage:')
print(data['Outcome'].value_counts(normalize=True) * 100)
```

```python
# Calculate correlation between features and outcome

correlation_with_outcome = data.corr()['Outcome'].sort_values(ascending=False)

print('Correlation of features with Outcome:')
print(correlation_with_outcome)
```

```python
# Calculate mean values of features grouped by Outcome

grouped_stats = data.groupby('Outcome').mean()

print('Mean values of features grouped by Outcome:')
print(grouped_stats)
```

```python
# Find the average glucose level for diabetic and non-diabetic patients

avg_glucose_diabetic = data[data['Outcome'] == 1]['Glucose'].mean()
avg_glucose_non_diabetic = data[data['Outcome'] == 0]['Glucose'].mean()

print(f'Average Glucose level for Non-Diabetic patients: {avg_glucose_non_diabetic:.2f}')
print(f'Average Glucose level for Diabetic patients: {avg_glucose_diabetic:.2f}')
```

```python
# Find the average BMI for diabetic and non-diabetic patients

avg_bmi_diabetic = data[data['Outcome'] == 1]['BMI'].mean()
avg_bmi_non_diabetic = data[data['Outcome'] == 0]['BMI'].mean()

print(f'Average BMI for Non-Diabetic patients: {avg_bmi_non_diabetic:.2f}')
print(f'Average BMI for Diabetic patients: {avg_bmi_diabetic:.2f}')
```

```python
# 8. Bar plot: Average feature values by Outcome

fig, axes = plt.subplots(2, 3, figsize=(15, 10))
fig.suptitle('Average Feature Values by Diabetes Outcome', fontsize=16)

features = ['Glucose', 'BloodPressure', 'SkinThickness', 'BMI', 'Age', 'Pregnancies']

for idx, feature in enumerate(features):
    ax = axes[idx // 3, idx % 3]
    avg_vals = data.groupby('Outcome')[feature].mean()
    sb.barplot(x=['Non-Diabetic', 'Diabetic'], y=avg_vals.values, hue=['Non-Diabetic', 'Diabetic'], 
               palette='Set2', ax=ax, legend=False)
    ax.set_title(f'Average {feature}')
    ax.set_ylabel(feature)
    ax.set_xlabel('')

plt.tight_layout()
plt.show()
```

```python
# Part 5 - Machine Learning Model Building
```

```python
# Import necessary ML libraries

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
# Split the data into X and Y, where X is input independent columns and Y is output dependent column

X = data.drop('Outcome', axis=1)
Y = data['Outcome']
```

```python
# Split the data into training and testing sets

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
# Create and train the Logistic Regression model

model = LogisticRegression(max_iter=1000)

model.fit(X_train, Y_train)
```

```python
# Calculate accuracy and display confusion matrix

Y_pred = model.predict(X_test)
accuracy = accuracy_score(Y_test, Y_pred)
print(f"Model Accuracy: {accuracy * 100:.2f}%")
print()

# Confusion matrix
conf_matrix = confusion_matrix(Y_test, Y_pred)
print("Confusion Matrix:")
print(conf_matrix)
```

```python
# Display classification report

print("Classification Report:")
print(classification_report(Y_test, Y_pred, target_names=['Non-Diabetic', 'Diabetic']))
```

```python
# Visualize the confusion matrix

plt.figure(figsize=(8, 6))
sb.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues', 
           xticklabels=['Non-Diabetic', 'Diabetic'], 
           yticklabels=['Non-Diabetic', 'Diabetic'])
plt.title('Confusion Matrix - Logistic Regression')
plt.ylabel('True Label')
plt.xlabel('Predicted Label')
plt.show()
```
